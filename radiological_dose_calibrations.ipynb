```python
import re

def get_calibration_factor(track_file, sensor_counts):
    with open(track_file, 'r') as f:
        content = f.read()
        # Locate the "sum over" line in the PHITS track output
        match = re.search(r'#\s+sum over\s+([0-9\.E\+\-]+)', content)
        if match:
            ref_dose_psv = float(match.group(1))
            # Convert pSv/count to uSv/count
            usv_per_count = (ref_dose_psv / sensor_counts) / 1e6
            return usv_per_count
    return None

# Using your validated Am-Be data
pos2_counts = 2.90e-05
cal_factor = get_calibration_factor("reference_hp10.out", pos2_counts)
print(f"Calibration Factor: {cal_factor:.2f} uSv/count")
```
